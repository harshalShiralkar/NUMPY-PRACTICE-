## NUMPY INDEXING AND SLICING OF VECTORS 

```python
arr = np.arange(11,21)
arr  
```

```python
arr[1:4]# [:4 it will go upto 4 th index from 0]
# [:] 0 to last element 
```

```python
arr[1::2]# Skips into 2 steps 
```

```python
arr[4]
```

## BOOLEAN INDEXING 

```python
arr = np.arange(11,21)
arr
```

```python
bool_index=arr%2 == 0 # It helps to give even numbers 
bool_index
```

```python
arr=arr[bool_index]
arr
```
